Apply capping

```python
cap_values = {
    "annual_inc": df["annual_inc"].quantile(0.999),
    "tot_coll_amt": df["tot_coll_amt"].quantile(0.999)
}

for col, cap in cap_values.items():
    df[col] = df[col].clip(upper=cap)
```
